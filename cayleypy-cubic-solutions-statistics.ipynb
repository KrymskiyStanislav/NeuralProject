# What is about ? 

Compute statistics several precompute solutions. 

Look on dependence on the beam size.

Solutions were obtained by CayleyPy-cube, with the following params: 


    
    hd1 256 
    hd2 128
    nrd 1
    batch size  100000
    lr  0.001
    Epochs 500 
    Kmax 26

    


```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file I/O (e.g. pd.read_csv)
import re
# import csv
import matplotlib.pyplot as plt
# Input data files are available in the read-only "../input/" directory
# For example, running this (by clicking run or pressing Shift+Enter) will list all files under the input directory

import os
```

```python
def get_solutions_stat(df_submit, source,IX_inf = 0):
    list_lens = []
    for k in range(len(df_submit)):
        sol = df_submit['path'].iloc[k]
        ln = len( sol.split('.'))
        list_lens.append(ln)
    l = np.array(list_lens[100:])
    mask = l < 90 # Higher length - unsolved 
    n_solved = mask.sum() 
    n_total = len(l)
    percent_solved = np.round(n_solved/n_total*100,1 )
    df_report = pd.DataFrame()
    IX = IX_inf
    #df_report.loc[IX,'source']= source
    df_report.loc[IX,'count (solved)']=  n_solved
    df_report.loc[IX,'count (total)']=  n_total
    if (n_solved>0):
        median_len = np.median( l[mask])
        mean_len = np.mean( l[mask])
        df_report.loc[IX,'min length (solved)']=  np.min( l[mask])
        df_report.loc[IX,'max length (solved)']=  np.max( l[mask]) 
        df_report.loc[IX,'std length (solved)']=  np.round( np.std( l[mask]) , 2)
        df_report.loc[IX,'percent (solved)']= percent_solved
        df_report.loc[IX,'median length (solved)']= median_len
        df_report.loc[IX,'mean length (solved)']= np.round(mean_len,1)
        df_report.loc[IX,'percentile10 length (solved)']= np.percentile( l[mask], 10 )
        df_report.loc[IX,'percentile90 length (solved)']= np.percentile( l[mask], 90 )
    else:
        median_len = -1
        mean_len = -1
        df_report.loc[IX,'min length (solved)']=  -1
        df_report.loc[IX,'max length (solved)']=  -1 
        df_report.loc[IX,'std length (solved)']=  -1
        df_report.loc[IX,'percent (solved)']= -1
        df_report.loc[IX,'median length (solved)']= -1
        df_report.loc[IX,'mean length (solved)']= -1
        df_report.loc[IX,'percentile10 length (solved)']= -1
        df_report.loc[IX,'percentile90 length (solved)']= -1
    #df_report.to_csv('df_report.csv')
    return df_report, n_solved
    

fn = '/kaggle/input/cayleypy-submissions/IhesCube_beam2p8_full/submission (48).csv'
df_submit = pd.read_csv(fn, index_col = 'initial_state_id')
display(df_submit.head(2))
df_report,n_solved = get_solutions_stat(df_submit,'submission (48).csv')
df_report
```

```python
list_dirs = [
    '/kaggle/input/cayleypy-submissions/IhesCube_beam2p5_full',
    '/kaggle/input/cayleypy-submissions/IhesCube_beam2p6_full',
    '/kaggle/input/cayleypy-submissions/IhesCube_beam2p8_full',
    '/kaggle/input/cayleypy-submissions/IhesCube_beam2p10_full',
    '/kaggle/input/cayleypy-submissions/IhesCube_beam2p12_full',
    '/kaggle/input/cayleypy-submissions/IhesCube_beam2p14_full',
    '/kaggle/input/cayleypy-submissions/IhesCube_beam2p16_full',
    '/kaggle/input/cayleypy-submissions/IhesCube_beam2p18_full',
    '/kaggle/input/cayleypy-submissions/IhesCube_full_submits'
            ]
df_report_full = pd.DataFrame()
for dirname in list_dirs:
    str_inf0 = dirname.split('_')[1]
    print(str_inf0)
    i_file = 0
    list_of_failed_names = ['']
    for dirpath, dirnames, filenames in os.walk(dirname):
        for filename in filenames:
            fullname = os.path.join(dirpath, filename)
            df = pd.read_csv(fullname,index_col ='initial_state_id')
            # display(df.head(1))    
            str_inf_full = str_inf0 + ' '+ str(i_file)
            i_file +=1
            df_report, n_solved = get_solutions_stat(df,filename,str_inf_full)    
            
            df_report.loc[str_inf_full,'Beam'] = str_inf0
            if len(df_report_full) == 0:
                df_report_full = df_report.copy()
            else:
                df_report_full = pd.concat([df_report_full, df_report], axis = 0)
            # display(df_report_full)
display(df_report_full) 
```

```python
df_report_full.to_csv("df_report_full.csv")
```

```python
df_report_aggregated = df_report_full.groupby('Beam').mean().round(1)
df_report_aggregated = df_report_aggregated.sort_values(
    "Beam",
    key=lambda s: s.str.extract(r'(\d+)$')[0].astype("Int64"),
    na_position="last",   # or "first"
)
df_report_aggregated.to_csv('df_report_aggregated.csv')
df_report_aggregated
```
